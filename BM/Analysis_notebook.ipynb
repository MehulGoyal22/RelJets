```python
#!/usr/bin/env python3
# Compute BM similarity variable chi and normalized pressure p_norm from a FLASH/yt checkpoint

import numpy as np
import yt

def bm_scale_pressure_radius(
    path,                      # full path to sedov_hdf5_chk_XXXX
    n_bins=2048,               # radial profile bins (log radius)
    k=0,                       # external density power-law (rho ∝ r^{-k}); 0 = uniform
    rho_ambient_param=1.0,     # fallback ambient density if outer sampling fails
    p_ambient_floor=0.0,       # optional floor to subtract from p2
    outer_frac=0.95,           # fraction of domain radius used to sample upstream (outer shell)
    smooth_win=21              # moving-average window (odd) for smoothing log p before gradient
):
    """
    Returns:
      chi      : (N,) BM similarity coordinate (dimensionless)
      p_norm   : (N,) normalized pressure p / (rho1 * Gamma_sh**2) (dimensionless)
      r        : (N,) radius array used to build chi (code_length)
      Rsh      : scalar, shock radius estimate (code_length)
      Gamma_sh : scalar, shock Lorentz-factor estimate (ultra-rel jump)
      rho1, p1 : scalars, upstream ambient samples
      t_code   : scalar, simulation time (code units)
      mask_bm  : boolean mask you can use to restrict plotting (e.g., r<=Rsh & chi>=1)
    Notes:
      - Assumes ultra-relativistic jump for Gamma_sh: p2 - p1 ≈ (2/3)*Gamma_sh^2 * rho1  (c=1)
      - Uses unit-safe chi with ct ≈ Rsh (good in the ultra-rel limit).
    """
    ds = yt.load(path)
    t_code = float(ds.current_time.to("code_time"))
    ctr = ds.domain_center
    r_max = float(0.999*(ds.domain_width.min()/2.0))

    # Angle-averaged radial profiles
    sp = ds.sphere(ctr, r_max)
    prof = yt.create_profile(
        sp, ("index","radius"),
        fields=[("gas","pressure"), ("gas","density"), ("index","cell_volume")],
        weight_field="cell_volume",
        n_bins=n_bins,
        accumulation=False,
        fractional=False,
        logs={("index","radius"): True},
    )

    r   = prof.x.to("code_length").value
    p   = prof[("gas","pressure")].value
    rho = prof[("gas","density")].value
    vol = prof[("index","cell_volume")].value

    # Guard tiny/NaN
    p   = np.where(np.isfinite(p),   p,   0.0)
    rho = np.where(np.isfinite(rho), rho, 0.0)
    p   = np.clip(p, 1e-300, None)

    # Upstream (ambient) sampling from outer shell
    outer = r > (outer_frac * r_max)
    if np.count_nonzero(outer) >= 10:
        rho1 = float(np.median(rho[outer]))
        p1   = float(np.median(p[outer]))
    else:
        rho1 = float(rho_ambient_param)
        p1   = float(p_ambient_floor)

    # Locate shock radius R_sh robustly: steepest drop of log p
    lp = np.log(p)
    if smooth_win is None or smooth_win < 3 or lp.size < smooth_win:
        lp_s = lp
    else:
        # simple moving average smoothing (no SciPy)
        w = smooth_win | 1  # ensure odd
        kernel = np.ones(w, dtype=float) / w
        # reflect-pad to avoid edge bias
        pad = w // 2
        lp_pad = np.r_[lp[pad-1::-1], lp, lp[:-pad-1:-1]]
        lp_s = np.convolve(lp_pad, kernel, mode="valid")
        # now lp_s length matches lp
        lp_s = lp_s[:lp.size]

    dldr = np.gradient(lp_s, r)
    i_shock = int(np.argmin(dldr))
    Rsh = float(r[i_shock])

    # Post-shock p2 (avoid mixing across the discontinuity)
    dr = float(abs(r[min(i_shock+1, len(r)-1)] - r[i_shock]))
    win_inside = (r >= Rsh - 3*dr) & (r <= Rsh - 1*dr)
    if np.count_nonzero(win_inside) < 3:
        win_inside = (r >= Rsh - 2*dr) & (r <= Rsh - 1*dr)
    p2 = float(np.median(p[win_inside])) if np.count_nonzero(win_inside) else float(np.max(p[r <= Rsh]))

    # Estimate Gamma_sh from ultra-relativistic jump
    # p2 - p1 ≈ (2/3) * Gamma_sh^2 * rho1  (with c=1)
    num = max(p2 - p1, 0.0)
    if rho1 <= 0.0 or num <= 0.0:
        # If this fails, return NaNs for chi/p_norm so caller can decide
        Gamma_sh = np.nan
        chi      = np.full_like(r, np.nan, dtype=float)
        p_norm   = np.full_like(p, np.nan, dtype=float)
        mask_bm  = (r <= Rsh)
        return chi, p_norm, r, Rsh, Gamma_sh, rho1, p1, t_code, mask_bm

    Gamma_sh = float(np.sqrt(1.5 * num / rho1))

    # BM similarity variable (unit-safe: use ct ≈ Rsh)
    chi = 1.0 + 2.0*(4.0 - k)*(Gamma_sh**2) * (1.0 - r / max(Rsh, 1e-30))

    # Normalized pressure for BM comparison
    p_norm = p / (rho1 * Gamma_sh**2)

    # A simple mask you can reuse for plotting (typical BM layer just behind the shock)
    mask_bm = (r <= Rsh) & (chi >= 1.0)

    return chi, p_norm, r, Rsh, Gamma_sh, rho1, p1, t_code, mask_bm


# ---------------- example usage ----------------
if __name__ == "__main__":
    # Example: point this to your checkpoint file
    chk = 11
    base = "/home/mehul/RelBlastwave/BM/RelSim/Tmax5/T10/Sedov"
    path = f"{base}/sedov_hdf5_chk_{chk:04d}"

    chi, p_norm, r, Rsh, Gamma_sh, rho1, p1, t_code, mask_bm = bm_scale_pressure_radius(
        path,
        n_bins=2048,
        k=0,
        rho_ambient_param=1.0,
        p_ambient_floor=0.0,
        outer_frac=0.95,
        smooth_win=21
    )

    # Print quick diagnostics; you handle plotting/slicing as you like
    print(f"t = {t_code:.6e}  R_sh ≈ {Rsh:.6f}  Γ_sh ≈ {Gamma_sh:.3f}  (rho1={rho1:.3e}, p1={p1:.3e})")
    # Example: restrict to your preferred region later, e.g.
    # bm_region = mask_bm & (chi <= 5.0)
    # chi_plot  = chi[bm_region]
    # p_plot    = p_norm[bm_region]
```

```python
import os
import numpy as np
import yt
from yt import add_field
from yt.units import speed_of_light, cm

# ----------------------------
# Derived field: rho * gamma^2
# ----------------------------
def _rho_gamma2(field, data):
    v = data[("gas", "velocity_magnitude")]
    beta2 = np.minimum((v / speed_of_light)**2, 1.0 - 1e-12)
    gamma = 1.0 / np.sqrt(1.0 - beta2)
    return data[("gas", "density")] * gamma**2

add_field(
    ("gas", "rho_gamma2"),
    function=_rho_gamma2,
    units="g/cm**3",
    sampling_type="cell",
)

outdir = "acc_phase/velocity"
os.makedirs(outdir, exist_ok=True)

N_BINS = 128
R_MIN  = 0.0  # cm; will be converted to a YTQuantity

for i in range(0, 70):
    ds = yt.load(f"/home/mehul/RelBlastwave/BM/RelSim/Tmax5/32blks/acc_phase/sedov_hdf5_chk_{i:04d}")

    c = ds.domain_center

    # largest inscribed sphere
    half_width = 0.5 * ds.domain_width.min()
    R_MAX = half_width.in_units("cm")

    sph = ds.sphere(c, R_MAX)

    # NOTE: bin_fields is a list/tuple
    prof = yt.create_profile(
        data_source=sph,
        bin_fields=[("index", "radius")],
        fields=[("gas", "velocity_magnitude")],
        n_bins=N_BINS,
        weight_field=("gas", "velocity_magnitude"),
        extrema={("index", "radius"): (R_MIN * cm, R_MAX)},
        logs={("index", "radius"): False},
        units={("index", "radius"): "cm", ("gas", "velocity_magnitude"): "cm/s"},
    )

    # Extract arrays
    r_cm   = prof.x.to("cm").v
    velocity = prof[("gas", "velocity_magnitude")].to("cm/s").v
    gamma = 1.0 / np.sqrt(1.0 - (velocity)**2)  # velocity in cm/s

    finite = np.isfinite(velocity)
    print(
        f"Checkpoint {i:04d}: r in [{r_cm[finite].min():.3e}, {r_cm[finite].max():.3e}] cm | "
        f"<velocity>_v max = {np.nanmax(velocity):.6e} cm/s"
        f"gamma max = {np.nanmax(gamma):.6e}"
        # f"gamma_calc = {np.nanmax(1.0 / np.sqrt(1.0 - (velocity[np.nanargmax(velocity)])**2)):.6e}" 
        #print gamma calculated from velocity
    )

    # # CSV
    # np.savetxt(
    #     os.path.join(outdir, f"rho_gamma2_profile_{i:04d}.csv"),
    #     np.column_stack([r_cm, rho_g2]),
    #     delimiter=",",
    #     header="radius_cm, rho_gamma2_velocity_weighted_g_per_cm3",
    #     comments=""
    # )

    # Plot
    import matplotlib.pyplot as plt
    plt.figure()
    plt.plot(r_cm, gamma)
    plt.yscale("log")
    plt.xscale("log")
    plt.xlabel("Radius (cm)")
    plt.ylabel(r"Velocity-weighted $\langle v \rangle_v$ (cm/s)")
    plt.title(f"Spherical average (vel-weighted) — chk {i:04d}")
    plt.text(
        0.15, -0.08,
        "No. of bins: " + str(N_BINS) + "\n" + r"$R_{max}=$" + f"{R_MAX:.2e} cm",
        ha="left", fontsize=10
    )
    plt.tight_layout()
    plt.savefig(os.path.join(outdir, f"gamma_profile_{i:04d}.png"), dpi=150)
    plt.close()
```

```python
import os
import numpy as np
import yt
from yt.units import speed_of_light, cm
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
import csv

# ----------------------------
# Settings
# ----------------------------
CHK_DIR   = "/home/mehul/RelBlastwave/BM/RelSim/Tmax5/32blks/acc_phase"
CHK_RANGE = range(1, 70)          # inclusive start, exclusive stop
N_BINS    = 128                   # resolution for radial profile
R_MIN_CM  = 0.0                   # cm
OUTDIR    = "acc_phase/gamma_peak_tracking"
WEIGHT    = ("gas", "velocity_magnitude")  # keep your velocity-weighted average; or set to None for volume-weighted

os.makedirs(OUTDIR, exist_ok=True)

times     = []
r_peaks   = []
g_peaks   = []
idx_list  = []

for i in CHK_RANGE:
    ds = yt.load(os.path.join(CHK_DIR, f"sedov_hdf5_chk_{i:04d}"))
    t_code = ds.current_time.to("code_time").value
    idx_list.append(i)
    times.append(t_code)

    # largest inscribed sphere (so bins cover full radial extent)
    c = ds.domain_center
    half_width = 0.5 * ds.domain_width.min()
    R_MAX = half_width.in_units("cm")

    sph = ds.sphere(c, R_MAX)

    prof = yt.create_profile(
        data_source=sph,
        bin_fields=[("index", "radius")],
        fields=[("gas", "velocity_magnitude")],
        n_bins=N_BINS,
        weight_field=WEIGHT,  # None for volume-weighted; your choice
        extrema={("index", "radius"): (R_MIN_CM * cm, R_MAX)},
        logs={("index", "radius"): False},
        units={
            ("index", "radius"): "cm",
            ("gas", "velocity_magnitude"): "cm/s",
        },
    )

    r_cm = prof.x.to("cm").v
    v    = prof[("gas", "velocity_magnitude")].to("cm/s").v

    # Guard against tiny numerical > c
    beta2 = np.minimum((v)**2, 1.0 - 1e-12)
    gamma = 1.0 / np.sqrt(1.0 - beta2)

    finite = np.isfinite(r_cm) & np.isfinite(gamma)
    r_cm   = r_cm[finite]
    gamma  = gamma[finite]

    # Peak finding on gamma(r): pick tallest peak; fallback to global max
    peaks, props = find_peaks(gamma, prominence=None, width=None)
    if peaks.size > 0:
        tallest = peaks[np.argmax(gamma[peaks])]
        r_pk = r_cm[tallest]
        g_pk = gamma[tallest]
    else:
        amax = np.nanargmax(gamma)
        r_pk = r_cm[amax]
        g_pk = gamma[amax]

    r_peaks.append(r_pk)
    g_peaks.append(g_pk)

    # Optional: quick per-chk debug print
    print(f"chk {i:04d}: t={t_code:.3e}, r_peak={r_pk:.6e} cm, gamma_peak={g_pk:.6e}")

# ----------------------------
# Save a CSV of the track
# ----------------------------
csv_path = os.path.join(OUTDIR, "gamma_peak_track.csv")
with open(csv_path, "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["checkpoint", "t_code", "r_peak_cm", "gamma_peak"])
    for i, t, rpk, gpk in zip(idx_list, times, r_peaks, g_peaks):
        w.writerow([i, f"{t:.8e}", f"{rpk:.8e}", f"{gpk:.8e}"])
print(f"Saved: {csv_path}")

# ----------------------------
# Plots vs time
# ----------------------------
# gamma_peak(t)
plt.figure(figsize=(7.4, 5.0))
plt.plot(times, g_peaks, marker="o", ms=3, lw=1)
plt.xlabel("t (code time)")
plt.ylabel(r"$\gamma_{\rm peak}$")
plt.title(r"Peak Lorentz factor vs time")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTDIR, "gamma_peak_vs_time.png"), dpi=160)
plt.close()

# r_peak(t)
plt.figure(figsize=(7.4, 5.0))
plt.plot(times, r_peaks, marker="o", ms=3, lw=1)
plt.xscale("log")
plt.yscale("log")
plt.xlabel("t (code time)")
plt.ylabel(r"$r_{\rm peak}$ (cm)")
plt.title(r"Radius of peak $\gamma$ vs time")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTDIR, "r_peak_vs_time.png"), dpi=160)
plt.close()

# ----------------------------
# gamma_peak vs r_peak
# ----------------------------
plt.figure(figsize=(7.4, 5.0))
plt.plot(r_peaks, g_peaks, marker="o", ms=3, lw=1)
plt.xlabel(r"$r_{\rm peak}$ (cm)")
plt.ylabel(r"$\gamma_{\rm peak}$")
plt.title(r"$\gamma_{\rm peak}$ vs $r_{\rm peak}$")

# optional: log x if radii span orders of magnitude
# rp = np.array(r_peaks)
# if np.isfinite(rp).all() and rp.max() / max(rp.min(), 1e-99) > 50:
#     plt.xscale("log")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTDIR, "gamma_peak_vs_r_peak.png"), dpi=160)
plt.close()


#linear fit

# ----------------------------
# gamma_peak vs r_peak + linear fit
# ----------------------------
import numpy as np
import matplotlib.pyplot as plt


r_arr = np.array(r_peaks)
g_arr = np.array(g_peaks)

mask = np.isfinite(r_arr) & np.isfinite(g_arr)
r_arr = r_arr[mask]
g_arr = g_arr[mask]

# Linear fit with covariance
coeffs, cov = np.polyfit(r_arr, g_arr, 1, cov=True)
a, b = coeffs
a_err, b_err = np.sqrt(np.diag(cov))

print(f"Linear fit: gamma = ({a:.6e} ± {a_err:.6e}) * r + ({b:.6e} ± {b_err:.6e})")

# Fit line for plotting
g_fit = np.polyval(coeffs, r_arr)

# Plot
plt.figure(figsize=(7.4,5.0))
plt.scatter(r_arr, g_arr, s=20, label="data", alpha=0.7)
plt.plot(r_arr, g_fit, "r-", lw=2,
         label=fr"fit: $\gamma = ({a:.2e}\pm{a_err:.2e})\,r + ({b:.2e}\pm{b_err:.2e})$")
plt.xlabel(r"$r_{\rm peak}$ (cm)")
plt.ylabel(r"$\gamma_{\rm peak}$")
plt.title(r"$\gamma_{\rm peak}$ vs $r_{\rm peak}$ (linear fit)")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(OUTDIR, "gamma_peak_vs_r_peak_fit.png"), dpi=160)
plt.close()



print("Wrote gamma_peak_vs_time.png and r_peak_vs_time.png")
```
